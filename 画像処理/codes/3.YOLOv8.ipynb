```python
# YOLOv8を読み込み（ultralyticsライブラリが提供）
from ultralytics import YOLO
```

```python
# 事前学習済みYOLOv8-nanoを読み込み（-cls.ptはImageNet事前学習で分類用）
model = YOLO('yolov8n-cls.pt')
```

```python
# モデルをCIFAR-10データで学習
# data="cifar10": 使用するデータセット
# epochs=3: 全データを3周
# imgsz=32: CIFAR-10画像を32x32サイズで処理
result = model.train(data="cifar10", epochs=3, imgsz=32)
```

```python
# 画像ファイルパスを使って推論を実行
model('./sample_horse_image.png')
```

```python
# 学習したモデルをONNX形式でエクスポート（他环境で推論実行用）
model.export(format="onnx")
```
