# 第3章

【問題】キーボードから、 100 以上の数が入力されるまで、止まらず入力を求め、100以上の数が入力されたら止まるプログラムを作れ。

解答例

```
# 最初に強制的に1回入力させる方針
x = int(input("x?"))
while x < 100:
    x= int(input("x?"))
```

```
# 最初に条件が成立しない値を入力しておく方式
key = 0
while key < 100:
    key = int(input("key?"))
```

```
# このプログラムはオマケ付き
key = int(input("key?"))

while key < 100:
    key = int(input("key?"))

print("end.")
```
