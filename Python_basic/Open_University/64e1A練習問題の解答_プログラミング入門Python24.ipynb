## フィボナッチ数列

フィボナッチ数列とは、次の漸化式によって表される数列$\{F_n\}$ のことである。

* $F_0 = 1 $
* $F_1 = 1 $
* $F_{k}= F_{k-1} + F_{k-2} $ （$k=2, 3, \cdots $のとき）

$n$　を入力させ、フィボナッチ数列の $F_{n} $ を求めるプログラムを作れ。

解答例

```
def fib(n):
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)

n = int(input("n?"))
print(fib(n))
```
