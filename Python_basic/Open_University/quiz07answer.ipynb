**QUIZ7-3**

pが正の整数のとき、pが素数かどうかを判定する関数を定義するプログラムを次に示す。

```python
def sosuka(p):
    flag = 1
    i = 2
    while i < p:
        r = p % i
        if r == 0:
            flag = 0
        i += 1
    return(flag)
```

次の選択肢のうち、正しいものを1つ選べ。

①7行目を「flag = 1」と書き換えると、pの値がどんな数でも、この関数の値は 1 になる。

②sosuka(10)を求めようとすると、5行目は10回実行される。

③sosuka(-2)の値は -2 となる。

④pが素数ならば、この関数の値は0になる。

正解は①です。flagの変数に代入しているところは、2行目と7行目のみで、どちらも1を代入しています。

```python
import random as rd

n = 300
incount = 0
for _ in range(n):
    x = rd.random()
    if  x  < 1 / 3 :
        incount += 1
print("kekka =", incount / n)
```
