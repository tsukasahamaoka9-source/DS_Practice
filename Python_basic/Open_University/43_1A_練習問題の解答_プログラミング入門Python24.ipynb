# 第4章

【問題】1 から 100 までの数のうち、7 の倍数をすべて表示するプログラムを作れ。

解答例

100が7の倍数でないことは暗算でもわかりますが、そのことは使わないことにして、プログラムを作ります。

次の例は、ｘを最初7にして、それに7を次々と加えていき、100を超えたら繰り返しをやめる、というプログラムである。

```
x = 7
while x <= 100:
    print(x)
    x += 7
```

次の例は、1から100までの整数をすべて列挙して、7の倍数なら表示、そうでないなら何もしない、というプログラムである。

```
for n in range(1, 101):
    if n % 7 == 0:
        print(n)
```

次の例は、 range に引数を3つ渡す方法で、目的を達成するプログラムである。

```
for m in range(7, 101, 7):
    print(m)
```
