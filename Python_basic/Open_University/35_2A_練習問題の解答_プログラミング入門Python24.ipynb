# 第3章

【問題】キーボードから、 "a"　か "b" 入力されるまで、止まらず入力を求め、"a" か "b" が入力されたら止まるプログラムを作れ。

解答例

```
key = input("key?")

while not(key == "a") and not(key == "b"):
    key = input("key?")

print("end.")
```

以下も正解となる。

```
key = input("key?")

while not(key == "a" or key == "b"):
    key = input("key?")

print("end.")
```

```
# 最初に、条件が成立しない初期値を与えておく方針
key = "c"

while not(key == "a" or key == "b"):
    key = input("key?")

print("end.")
```
