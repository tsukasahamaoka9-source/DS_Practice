**QUIZ3-2**

次のプログラムを動作させた。

```python
x = int(input('x?'))

if x > 50:
    print('A')
elif x > 25:
    print('B')
else:
    print('C')
```

このとき、B が出力される入力 x の値として正しいもの選べ。（実行せずに解答せよ。）

① 15

② 45

③ 65

④ 75

正解は②です。 x が 45 のときは、 x > 50 の条件が成り立たないため、その次の elif x > 25 を判定します。成り立つので、 B を表示して、その後は何も表示しないでプログラムは終了します。

```python
x = 45

if x > 50:
    print('A')
elif x > 25:
    print('B')
else:
    print('C')
```

**QUIZ3-4**

次のプログラムを動作させた。

```python
nishi = 2

def kazoeru(x):
    higashi = x
    return(nishi + higashi)

print(kazoeru(3))
```

このとき、次の選択肢のうち、正しいものを1つ選べ。

① 変数 higashi は大域変数である。

② プログラム終了時に、nishi の値は 5 である。

③ プログラム終了後に、print(higashi) を実行するとエラーになる。

④ 出力される結果は 10 である。

正解は③です。変数 higashi は局所変数であり、関数の中で呼び出すことだけ許されます。したがって、このプログラムを実行したあとに print(higashi)  を実行するとエラーが表示されます。

```python
nishi = 2

def kazoeru(x):
    higashi = x
    return(nishi + higashi)

print(kazoeru(3))
print(higashi)
```

**QUIZ3-5**

次のプログラムを動作させた。

```python
i = 1
n = 10
while i < n:
    print(i, end="")
    i += 4
```

このとき、次の選択肢のうち、出力結果として正しいものを1つ選べ。（実行せずに解答せよ。）

① 159

② 15913

③ 04812

④ 012345678910

正解は①です。while があるので条件が成立している間繰り返しが行われます。その間に、変数 i は、 1, 5, 9, 13 と変化します。i が 13 になっているときに while の条件が成立しなくなるので、繰り返しが止まります。
