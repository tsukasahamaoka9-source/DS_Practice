## 練習問題

【問題】次の通りに表示するプログラムを、その下に作れ。

Q1. AkaAoKiiro

解答

```
s1 = "Aka"
s2 = "Ao"
s3 = "Kiiro"
print(s1 + s2 + s3)
```

Q2. Aka Ao Kiiro

解答

```
s1 = "Aka"
s2 = "Ao"
s3 = "Kiiro"
print(s1, s2, s3)
```

Q3. Aka, Ao, Kiiro

解答

```
s1 = "Aka"
s2 = "Ao"
s3 = "Kiiro"
print(s1 + ", " + s2 + ", " + s3)
```

```
s1 = "Aka"
s2 = "Ao"
s3 = "Kiiro"
print(s1 + ",", s2 + ",", s3)
```
