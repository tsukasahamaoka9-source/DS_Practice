## 文字列の結合
【問題】文字列変数 m　に、シングルクォーテーションマークと、ダブルクォーテーションマークを共に入れて、

シングルクォーテーションは ' です。ダブルクォーテーションマークは " です。

と表示させる方法を考えよ。

解答

```
a = "シングルクォーテーションマークは ' です。"
b = 'ダブルクォーテーションマークは " です。'
m = a + b
print(m)
```
