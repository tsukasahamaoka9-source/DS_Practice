**QUIZ4-1**

次の出力を得たいと考えて、下のプログラムを作成した。

得たい出力
```
アイスコーヒー
ミルクコーヒー
ホットコーヒー
アイスティ
ミルクティ
ホットティ
```

作成したプログラム

````python
drink = ("コーヒー", "ティ")
addon = ("アイス", "ミルク", "ホット")

for d in drink:
    for a in addon:
        print(d + a)
````

だが、「得たい出力」を得ることができなかった。
プログラムをどのように修正すればよいか。正しいものを次の選択肢から1つ選べ。（実行せずに解答せよ。）

①1行目と2行目を入れ替える。

②3行目だけを削除して、全体として1行減らす。

③4行目を for a in addon: にして、5行目を for  d in drink: にする。

④5行目を print(a + d) にする。

正解は④です。上のプログラムをそのまま実行すると、次の出力になります。
```
コーヒーアイス
コーヒーミルク
コーヒーホット
ティアイス
ティミルク
ティホット
```
得たい出力と違っているところは、「drink」が先（左）にあり、「addon」が後（右）にあることです。そこを直す必要があります。

````python
drink = ("コーヒー", "ティ")
addon = ("アイス", "ミルク", "ホット")

for d in drink:
    for a in addon:
        print(a + d)
````

**QUIZ4-2**

次のプログラムを作成して実行しようとした。

````python
for i in range(3):
    print("おはよう", i)
    print("おやすみ")
````

しかし、文法エラーが発生してしまった。このプログラムについて、次の選択肢のうち、正しいものを1つ選べ。

①1行目の range(3) を range(i) とすると文法エラーは発生しない。

②2行目の print("おはよう",i) を print("おやすみ", i) とすると文法エラーは発生しない。

③2行目の 左側のインデント（字下げ）を3行目と同じ4文字にすると文法エラーは発生しない。

④3行目の 左側のインデント（字下げ）を16文字にすると文法エラーは発生しない。

正解は③です。1行目のforは、直下にブロックを求めます。同じ幅のインデントはブロックを形成します。2行目と3行目のインデントは異なるので、2行目と3行目は異なるブロックです。1行目のforは2行目をブロックとみなします。しかし、3行目のブロックを求める記述はありません。そこで、ここを③の通りに修正することが必要です。他の選択肢は文法エラーを解消しません。

**QUIZ4-4**

我が国で発行されている紙幣の金額（単位は「円」）は、現在、10000, 5000, 2000, 1000 がある。金額を指定されたとき、その金額をすべて紙幣にすることを考える。このとき、なるべく紙幣の枚数を少なくしたい。そこで、次のプログラムを作成した。

````python
# 紙幣の金額を定義する
shiheilist = (1000, 2000, 5000, 10000)

# 金額を入力する
p = int(input("how much?"))

for shihei in shiheilist:
    print(shihei, " yen :", p // shihei, "mai")
    p = p % shihei
````

実行して、pとして23000円を入力したところ、

```
1000  yen : 23 mai
2000  yen : 0 mai
5000  yen : 0 mai
10000  yen : 0 mai
```

と表示され、期待した結果を得られなかった。上のプログラムを修正して、紙幣のみで構成できるどんな金額を入力しても、目的に合った適切な結果を表示させる正しい方法を次の選択肢から1つ選べ。

①2行目を「shiheilist = (10000, 5000, 2000, 1000)」とする。

②7行目を「for shiheilist in shihei:」とする。

③8行目と9行目を入れ替える。

④9行目を「p = p // shihei」とする。

正解は①です。紙幣の枚数を少なくするには、まず、金額が大きい硬貨から順に、使えるかどうかを確認するため、 shiheilist は降順にしておく必要があります。

````python
# 紙幣の金額を定義する
shiheilist = (10000, 5000, 2000, 1000)

# 金額を入力する
p = int(input("how much?"))

for shihei in shiheilist:
    print(shihei, " yen :", p // shihei, "mai")
    p = p % shihei
````

**QUIZ4-5**

次のプログラムを作成して実行しようとした。

````python
import random as rd # 乱数を発生させる関数ライブラリを取り込む

x = rd.randint(0, 8)
r = 7 / x
print(r)
````

何回か実行したところ、値が表示されることもあれば、エラーになることもあった。なぜエラーになるのか、正しい説明を次の選択肢から1つ選べ。

①2行目が空行だから。

②3行目で x が 0 になることがあるから。

③4行目の 7 / x が割り切れないことがあるから。

④5行目が字下げされていないから。

正解は②です。randint(s, e) は、random ライブラリを必要とし、また、s以上e以下の整数を乱数として生成します。元のプログラムの3行目では x に 0 を代入することがあ、そのとき4行目で 0で割ることになりますが、それはできないため、エラーになります。
