**QUIZ6-1**

選択ソートで長さ n のリストを整列させるときについて、次の選択肢のうち、正しいものを1つ選べ。

① 長さが 4 のリストを整列させるには、要素の大小関係の比較を 4 回行う。

② 長さが 5 のリストを整列させるには、要素の大小関係の比較を 10 回行う。

③ 長さが 6 のリストを整列させるには、要素の大小関係の比較を 18 回行う。

④ 長さが 7 のリストを整列させるには、要素の大小関係の比較を 22 回行う。

正解は②です。長さ n のリストを選択ソートで整列させるには、要素の大小関係の比較を n(n-1) /2  回行います。



n=5 を代入すると 5 * 4 / 2 = 10 回となります。

```python
for n in range(4, 8):
    print("n =", n, ": n(n-1)/2 =", n*(n-1) / 2)
```

**QUIZ6-2**

ユーザが指定した要素kが、二分探索で長さ n のリストの中にあるかどうかを調べる手順について、次の選択肢のうち、正しいものを1つ選べ。なお、長さ1のリストに対しては、比較を1回行うとする。

① 長さが 5 のリストから二分探索するには、要素kとの比較を2回行う。

② 長さが 6 のリストから二分探索するには、要素kとの比較を3回行う。

③ 長さが 7 のリストから二分探索するには、要素kとの比較を9回行う。

④ 長さが 9 のリストから二分探索するには、要素kとの比較を13回行う。

正解は②です。1回の比較で探索範囲は半分になります。長さ6のリストでは、6, 3, 1 となり、3回の比較が必要です。

```python
nlist = (5, 6, 7, 9) # nlist は、長さのリスト
for n in nlist:
    kaisu = 0
    print("n =", n, end=": ")
    m = n
    while m > 0:
        print(m, end=" ")
        m = int( m / 2)
        kaisu += 1
    print("; kaisu =", kaisu)
```

**QUIZ6-3**

S(n) は、1からnまでの数の和である。S(n) を求めるために、次の関数 nsum を作成した。

```python
def nsum(k):
   if k == 0:
       return(0)
   else:
       return(k + nsum(k - 1))
```

次の選択肢のうち、正しいものを1つ選べ。

①nsum(0)を求めるとき、2行目の判断は行われない。

②nsum(1)を求めるとき、2行目の判断はちょうど2回行われる。

③nsum(2)を求めるとき、2行目の判断はちょうど4回行われる。

④nsum(3)を求めるとき、2行目の判断はちょうど6回行われる。

正解は②です。nsum(1)を求めるときは、2行目の判断は2回行われます。

```python
def nsum(k):
    print("*", end="") # 元 2行目が呼ばれたら * を1つ表示する
    if k == 0: # 元 2行目
        return(0)
    else:
        return(k + nsum(k - 1))

for n in range(4):
    print("n =", n, end=": ")
    nsum(n)
    print("")
```
