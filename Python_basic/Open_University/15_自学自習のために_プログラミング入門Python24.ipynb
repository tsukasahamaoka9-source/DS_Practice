# 1.5 自学自習のために

## もなかこみフォントの利用（オプション）

プログラムを作っていると、全角文字と半角文字の違いがわかりにくくて、プログラムが間違っているのに、それを見つけるために時間がかかることがある。

例えば、次の2つのプログラム、左側はエラーがないが、右側の print の中の右側のダブルクォーテーションマークが全角文字になっているため、エラーとなる。

スクリーンショット 2023-01-09 19.44.39.png

そこで、この違いがわかりやすいフォントを利用することで、間違いを早く発見できるようにする事が可能である。

ここで紹介する「もなかこみフォント」は、アシアル情報教育研究所が開発したオープンライセンス（無料で利用可能）なフォントである。

間違いやすい文字の全角文字の場合は、細い四角枠で囲むようになっている。

https://anko.education/tool/monacakomi

インストール方法は、上記のページに従ってフォントをダウンロードしてパソコンにインストールし、その後、ブラウザの「固定幅フォント」として、このフォントを指定するだけである。Chrome の場合だと以下の通り。

1. ブラウザの右上にあるケバブメニュー（点が縦方向に3つ並んでいる）から、「設定」を選ぶ。
2. 左側の「デザイン」タブを選ぶ。
3. 下の方の「固定幅フォント」のフォント名をプルダウンする。
4. 一覧（ABC順になっていない）から monacakomi を探して指定する。
