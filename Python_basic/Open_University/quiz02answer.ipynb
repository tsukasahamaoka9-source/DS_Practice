```python
#①
t = int(input("a?"))
print(t + 1)
```

```python
#②
t = input("a?")
print(t + 1)
```

**QUIZ2-4**

keys = "abcdefgh" とする。次の選択肢のうち、誤っているものを1つ選べ。

① keys[0:1] は、 "a" である。

② keys[2:2] は、 空列（長さ0の文字列）である。

③ keys[0:100] は、 "abcdefgh" である。

④ keys[10:0] は、 "hgfedcba" である。

正解は④です。 もし、文字を逆向きにしたいときは、 keys[::-1]と書きます。

```python
# 確認用
keys = "abcdefgh"
print(keys[0:1]) # は、 "a" である。
print(keys[2:2]) # は、 空列（長さ0の文字列）である。
print(keys[0:100]) # は、 "abcdefgh" である。
print(keys[10:0]) # は、 "hgfedcba" である。
```
