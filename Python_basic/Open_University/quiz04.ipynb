**QUIZ4-4**

我が国で発行されている紙幣の金額（単位は「円」）は、現在、10000, 5000, 2000, 1000 がある。金額を指定されたとき、その金額をすべて紙幣にすることを考える。このとき、なるべく紙幣の枚数を少なくしたい。そこで、次のプログラムを作成した。

````python
# 紙幣の金額を定義する
shiheilist = (1000, 2000, 5000, 10000)

# 金額を入力する
p = int(input("how much?"))

for shihei in shiheilist:
    print(shihei, " yen :", p // shihei, "mai")
    p = p % shihei
````

実行して、pとして23000円を入力したところ、

```
1000  yen : 23 mai
2000  yen : 0 mai
5000  yen : 0 mai
10000  yen : 0 mai
```

と表示され、期待した結果を得られなかった。上のプログラムを修正して、紙幣のみで構成できるどんな金額を入力しても、目的に合った適切な結果を表示させる正しい方法を次の選択肢から1つ選べ。

①2行目を「shiheilist = (10000, 5000, 2000, 1000)」とする。

②7行目を「for shiheilist in shihei:」とする。

③8行目と9行目を入れ替える。

④9行目を「p = p // shihei」とする。
